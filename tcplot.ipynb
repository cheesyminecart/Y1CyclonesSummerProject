```python
# tc plots. done in notebook to allow for greater upscaling during this initial phase of our project!
# using this tutorial: https://fabienmaussion.info/climate_system/week_02/01_Lesson_NetCDF_Data.html
# isabel1.nc testfile (sorry!)

import matplotlib.pyplot as plt  # plotting library
import numpy as np  # numerical library
import xarray as xr  # netCDF library
import cartopy  # Map projections libary
import cartopy.crs as ccrs  # Projections list 

ds = xr.open_dataset(r'/Users/juliencheung/Documents/GitHub/Y1CyclonesSummerProject/data/isabel2.nc') #peak intensity file, we should aim for a strengthening, peak, then weakening image for each cyclone.
ds

u = ds.u10.sel(valid_time="2003-09-11T18:00:00", method='nearest').load()
v = ds.v10.sel(valid_time="2003-09-11T18:00:00", method='nearest').load()

magn = np.sqrt(u**2 + v**2) # do not use .values, that converts it into a numpy array and we can't do anything

ax = plt.axes(projection=ccrs.PlateCarree())
qv = ax.quiver(u.longitude, u.latitude, u, v, transform=ccrs.PlateCarree())

ax.coastlines(color='grey');

plt.show()

magn.plot.contourf()
```
