```python
# hurricane tammy cyclone plots

import matplotlib.pyplot as plt  # plotting library
import numpy as np  # numerical library
import xarray as xr  # netCDF library
import cartopy  # Map projections libary
import cartopy.crs as ccrs  # Projections list 

ds = xr.open_dataset(r'/Users/juliencheung/Documents/GitHub/Y1CyclonesSummerProject/data/isabel1.nc')
ds

u = ds.u10.sel(valid_time="2003-09-07T12:00:00", method='nearest').load()
v = ds.v10.sel(valid_time="2003-09-07T12:00:00", method='nearest').load()

ax = plt.axes(projection=ccrs.PlateCarree())  # note that I changed the map projection
qv = ax.quiver(u.longitude, u.latitude, u, v, transform=ccrs.PlateCarree())
ax.coastlines(color='grey');
```
